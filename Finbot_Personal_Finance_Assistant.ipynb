```python
import ipywidgets as widgets
from IPython.display import display, HTML
from datetime import datetime
import random

# ---------- FINBOT THEME ----------
display(HTML("""
<style>
.finbot-header {
    background: linear-gradient(135deg, #24124d, #7048e8);
    color: white;
    padding: 24px;
    border-radius: 18px;
    text-align: center;
    font-family: Arial;
    margin-bottom: 12px;
}
.finbot-header h1 {
    color: white;
    margin: 0;
    font-size: 30px;
}
.finbot-header p {
    margin-bottom: 0;
    color: #eee8ff;
}
</style>
"""))

header = widgets.HTML("""
<div class="finbot-header">
    <h1>💜 FinBot</h1>
    <p>Your friendly Personal Finance Assistant</p>
    <p>✨ Learn • Save • Budget • Grow</p>
</div>
""")

chat = widgets.Output(
    layout=widgets.Layout(
        height="330px",
        overflow_y="auto",
        border="1px solid #ded7f5",
        padding="12px",
        width="100%"
    )
)

# ---------- CHAT RESPONSES ----------
def get_response(message):
    msg = message.lower()

    if any(word in msg for word in ["hello", "hi", "hey"]):
        return "Hey there! 👋 Welcome to FinBot. How can I help with your money goals today?"

    elif any(word in msg for word in ["budget", "expenses", "spending"]):
        return ("📊 **Budget Tip**<br>"
                "Try the 50/30/20 rule as a starting point:<br>"
                "🟢 50% for needs<br>"
                "💜 30% for wants<br>"
                "💰 20% for savings and debt repayment.<br>"
                "Adjust these percentages to suit your income and essential expenses.")

    elif any(word in msg for word in ["save", "saving", "savings"]):
        return ("💰 **Smart Saving Tips**<br>"
                "1. Set a monthly savings goal.<br>"
                "2. Track small daily expenses.<br>"
                "3. Save automatically when possible.<br>"
                "4. Build an emergency fund gradually.")

    elif any(word in msg for word in ["invest", "investment", "stocks", "mutual fund"]):
        return ("📈 **Investing Basics**<br>"
                "Learn about risk, diversification, fees and time horizon first. "
                "Investments can lose value, and returns are not guaranteed. "
                "Choose options based on your goals and risk tolerance.")

    elif any(word in msg for word in ["debt", "loan", "credit card"]):
        return ("💳 **Debt Management**<br>"
                "List your debts, interest rates and minimum payments. "
                "Pay required minimums on time, then consider prioritising "
                "high-interest debt.")

    elif any(word in msg for word in ["goal", "emergency", "fund"]):
        return ("🎯 **Your Financial Goal**<br>"
                "Start with a realistic target and a deadline. "
                "For an emergency fund, build toward several months "
                "of essential expenses as your circumstances allow.")

    elif any(word in msg for word in ["help", "options", "menu"]):
        return ("✨ **I can help you with:**<br>"
                "💰 Saving money<br>📊 Budgeting<br>"
                "📈 Investing basics<br>💳 Debt management<br>"
                "🎯 Financial goals")

    else:
        return ("😊 I'm still learning! Try asking about budgeting, "
                "saving, investing basics, debt or financial goals.")

# ---------- DISPLAY MESSAGES ----------
def add_message(sender, message, color):
    with chat:
        display(HTML(f"""
        <div style="background:{color}; padding:12px;
        margin:8px 0; border-radius:12px;
        font-family:Arial; line-height:1.6;
        overflow-wrap:anywhere;">
        <b>{sender}</b><br>{message}
        </div>
        """))

# ---------- SEND MESSAGE ----------
def send_message(button=None):
    message = text_box.value.strip()

    if not message:
        return

    safe_message = (message.replace("&", "&amp;")
                    .replace("<", "&lt;")
                    .replace(">", "&gt;"))

    add_message("🧑 You", safe_message, "#eee8ff")
    text_box.value = ""

    response = get_response(message)
    add_message("🤖 FinBot", response, "#e9f8f0")

# ---------- INPUT AREA ----------
text_box = widgets.Text(
    placeholder="Ask me about saving, budgeting...",
    description="",
    layout=widgets.Layout(width="78%")
)

send_button = widgets.Button(
    description="Send ➤",
    button_style="primary",
    layout=widgets.Layout(width="20%", height="36px")
)

send_button.on_click(send_message)
text_box.on_submit(send_message)

# ---------- QUICK ACTIONS ----------
def quick_question(question):
    text_box.value = question
    send_message()

quick_buttons = widgets.HBox([
    widgets.Button(description="💰 Saving"),
    widgets.Button(description="📊 Budget"),
    widgets.Button(description="📈 Investing"),
    widgets.Button(description="💳 Debt")
], layout=widgets.Layout(
    flex_flow="row wrap",
    gap="6px"
))

for button, question in zip(
    quick_buttons.children,
    ["How can I save money?",
     "Give me budgeting tips",
     "Explain investing basics",
     "How can I manage debt?"]
):
    button.on_click(lambda b, q=question: quick_question(q))

# ---------- LAUNCH APP ----------
display(header)
display(chat)
display(quick_buttons)
display(widgets.HBox([text_box, send_button]))

add_message(
    "🤖 FinBot",
    "Welcome! 💜 I'm your personal finance learning assistant. "
    "Choose a quick topic or type a question below. "
    "Type <b>help</b> to see what I can do.",
    "#e9f8f0"
)
```
